```python
import pandas as pd
import altair as alt
```

```python
df = pd.read_csv("valenzuela_database_limpia (1).csv")

df.head()
```

```python
df_limpia = df.copy()

# Revisar columna año y dejar en formato númerico
df_limpia["Año"] = pd.to_numeric(df_limpia["Año"], errors="coerce")

# Convertir tasa global de fecundidad
df_limpia["Tasa global Fecundidad"] = (
    df_limpia["Tasa global Fecundidad"]
    .astype(str)
    .str.replace(",",".", regex=False)
)
df_limpia["Tasa global Fecundidad"] = pd.to_numeric(
    df_limpia["Tasa global Fecundidad"], errors="coerce"
)

# Convertir tasa bruta de natalidad
df_limpia["Tasa bruta natalidad"] = (
    df_limpia["Tasa bruta natalidad"]
    .astype(str)
    .str.replace(",",".", regex=False)
)

df_limpia["Tasa bruta natalidad"] = pd.to_numeric(
    df_limpia["Tasa bruta natalidad"], errors="coerce"
)

# Ordenar por año
df_limpia = df_limpia.sort_values("Año")

df_limpia.head()
```

```python
# Gráfico 1: evolución de nacidos vivos

grafico_nacidos = alt.Chart(df_limpia).mark_line(point=True).encode(
    x=alt.X(
        "Año:O",
        title="Año"
    ),
    y=alt.Y(
        "Nacidos vivos:Q",
        title="Nacidos vivos"
    ),
    tooltip=[
        alt.Tooltip("Año:O", title="Año"),
        alt.Tooltip("Nacidos vivos:Q", title="Nacidos vivos", format=",")

    ]
).properties(
    title="Evolución de los nacidos vivos en Chile, 1992-2024",
    width=800,
    height=400
).interactive()

grafico_nacidos
```

```python
# Gráfico 2: evolución de la tasa global de fecundidad

grafico_fecundidad = alt.Chart(df_limpia).mark_line(point=True).encode(
    x=alt.X(
        "Año:O",
        title="Año"
    ),
    y=alt.Y(
        "Tasa global Fecundidad:Q",
        title="Tasa global de fecundidad"
    ),
    tooltip=[
        alt.Tooltip("Año:O", title="Año"),
        alt.Tooltip("Tasa global Fecundidad:Q", title="Tasa global de fecundidad", format=".2f")
    ]
).properties(
    title="Evolución de la tasa global de fecundidad en Chile, 1992-2024",
    width=800,
    height=400
).interactive()

grafico_fecundidad
```

```python
# Gráfico 3: comparación 1992 vs 2024

df_limpia["Nacidos vivos"] = (
    df_limpia["Nacidos vivos"]
    .astype(str)
    .str.replace(".","", regex=False)
    .str.replace(",",".", regex=False)
)

df_limpia["Nacidos vivos"] = pd.to_numeric(
    df_limpia["Nacidos vivos"],
    errors="coerce"
)

primer_ano = df_limpia["Año"].min()
ultimo_ano = df_limpia["Año"].max()

df_comparacion = df_limpia[df_limpia["Año"].isin([primer_ano, ultimo_ano])].copy()

df_comparacion_largo = df_comparacion.melt(
    id_vars=["Año"],
    value_vars=[
        "Nacidos vivos",
        "Tasa global Fecundidad",
        "Tasa bruta natalidad"
    ],
    var_name="Indicador",
    value_name="Valor"
)

df_comparacion_largo["Indicador"] = df_comparacion_largo["Indicador"].replace({
    "Nacidos vivos": "Nacidos vivos",
    "Tasa global Fecundidad": "Hijos promedio por mujer",
    "Tasa bruta natalidad": "Nacidos por cada 1.000 habitantes"
})

base = alt.Chart(df_comparacion_largo).encode(
    x=alt.X("Año:O", title="Año"),
    y=alt.Y("Valor:Q", title="Valor"),
    color=alt.Color(
        "Año:O",
        title="Año",
        scale=alt.Scale(
            domain=[primer_ano, ultimo_ano],
            range=["steelblue","crimson"]
        )
    ),
    tooltip=[
        alt.Tooltip("Indicador:N", title="Indicador"),
        alt.Tooltip("Año:O", title="Año"),
        alt.Tooltip("Valor:Q", title="Valor", format=",.2f")
    ]
).properties(
    width=220,
    height=300
)

barras = base.mark_bar()

textos = base.mark_text(
    dy=-8,
    fontSize=11
).encode(
    text=alt.Text("Valor:Q", format=",.2f")
)

grafico_antes_despues = alt.layer(
    barras,
    textos,
).facet(
    column=alt.Column("Indicador:N", title=None)
).properties(
    title="Cambio de los principales indicadores de natalidad en Chile entre 1992 y 2024"

).resolve_scale(
    y="independent"
)

grafico_antes_despues

```
