```python
import pandas as pd
import altair as alt
```

```python
df = pd.read_csv("jadue_database_limpia (2).csv", sep=";")
```

```python
df.head()
```

```python
df = df.drop(columns=["Unnamed: 6"])
```

```python
print(df.columns)
```

```python
df_teen = df[df["grupo-edad-madre"] == "15-19"]
```

```python
df_teen["tasa-especifica-fecundidad"] = (
    df_teen["tasa-especifica-fecundidad"]
    .astype(str)
    .str.replace(",", ".")
    .astype(float)
    )
```

```python
grafico = alt.Chart(df_teen).mark_line(point=True).encode(
  x='Año:O',
  y='tasa-especifica-fecundidad:Q'
).properties(
    title='Evolución de la fecundidad adolescente en Chile'
)

grafico
```

```python
grafico.save("grafico.html")
```
