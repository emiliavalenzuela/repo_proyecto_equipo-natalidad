```python
import pandas as pd
import altair as alt

df_ranking = df_limpia.copy()

df_ranking["ano"] = pd.to_numeric(df_ranking["ano"], errors="coerce")

df_ranking["tasa_natalidad_1000"] = (
    df_ranking["tasa_natalidad_1000"]
    .astype(str)
    .str.replace(",", ".", regex=False)
)

df_ranking["tasa_natalidad_1000"] = pd.to_numeric(
    df_ranking["tasa_natalidad_1000"],
    errors="coerce"
)

df_ranking = df_ranking.dropna(subset=["ano", "region", "tasa_natalidad_1000"])
```

```python
orden_norte_sur_todas = [
    "arica_parinacota",
    "tarapaca",
    "antofagasta",
    "atacama",
    "coquimbo",
    "valparaiso",
    "metropolitana",
    "ohiggins",
    "maule",
    "nuble",
    "biobio",
    "la_araucania",
    "los_rios",
    "los_lagos",
    "aysen",
    "magallanes"
]

orden_norte_sur_antiguas = [
    "tarapaca",
    "antofagasta",
    "atacama",
    "coquimbo",
    "valparaiso",
    "metropolitana",
    "ohiggins",
    "maule",
    "biobio",
    "la_araucania",
    "los_lagos",
    "aysen",
    "magallanes"
]
```

```python
df_1993 = df_ranking[
    (df_ranking["ano"] == 1993) &
    (df_ranking["region"].isin(orden_norte_sur_antiguas))
][["region", "tasa_natalidad_1000"]].rename(
    columns={"tasa_natalidad_1000": "tasa_1993"}
)

df_2022 = df_ranking[
    (df_ranking["ano"] == 2022) &
    (df_ranking["region"].isin(orden_norte_sur_antiguas))
][["region", "tasa_natalidad_1000"]].rename(
    columns={"tasa_natalidad_1000": "tasa_2022"}
)

rango_1993_2022 = df_1993.merge(df_2022, on="region")

rango_1993_2022["inicio_barra"] = rango_1993_2022[["tasa_1993", "tasa_2022"]].min(axis=1)
rango_1993_2022["fin_barra"] = rango_1993_2022[["tasa_1993", "tasa_2022"]].max(axis=1)
rango_1993_2022["cambio"] = rango_1993_2022["tasa_2022"] - rango_1993_2022["tasa_1993"]

grafico_rango_1993_2022 = alt.Chart(rango_1993_2022).mark_bar(size=18).encode(
    x=alt.X(
        "inicio_barra:Q",
        title="Tasa de natalidad por cada 1.000 habitantes"
    ),
    x2="fin_barra:Q",
    y=alt.Y(
        "region:N",
        sort=orden_norte_sur_antiguas,
        title="Región"
    ),
    tooltip=[
        alt.Tooltip("region:N", title="Región"),
        alt.Tooltip("tasa_1993:Q", title="Tasa 1993", format=".2f"),
        alt.Tooltip("tasa_2022:Q", title="Tasa 2022", format=".2f"),
        alt.Tooltip("cambio:Q", title="Cambio", format=".2f")
    ]
).properties(
    title="Cambio en la tasa de natalidad por región, 1993-2022",
    width=750,
    height=500
)

grafico_rango_1993_2022
```

```python
#grafico: proporción de nacimientos según origen de la madre por región
import pandas as pd
import altair as alt
df_proporcion = df_limpia.copy()
#convertir año a número
df_proporcion["ano"] = pd.to_numeric(df_proporcion["ano"], errors="coerce")
#convertir porcentaje de madres extranjeras a número
df_proporcion["porcentajes_madres_extranjeras"] = df_proporcion["porcentajes_madres_extranjeras"].astype(str).str.replace(",", ".", regex=False)
df_proporcion["porcentajes_madres_extranjeras"] = pd.to_numeric(df_proporcion["porcentajes_madres_extranjeras"], errors="coerce")
#dejar solo filas con datos completos
df_proporcion = df_proporcion.dropna(subset=["ano", "region", "porcentajes_madres_extranjeras"])
#crear porcentaje complementario
df_proporcion["porcentaje_madres_internas"] = 100 - df_proporcion["porcentajes_madres_extranjeras"]
#crear tabla con año inicial
df_inicio = df_proporcion[df_proporcion["ano"] == df_proporcion["ano"].min()]
#transformar a formato largo
df_proporcion_largo = df_proporcion.melt(id_vars=["ano", "region"], value_vars=["porcentajes_madres_extranjeras", "porcentaje_madres_internas"], var_name="origen_madre", value_name="porcentaje")
#renombrar categorías para que se entienda mejor
df_proporcion_largo["origen_madre"] = df_proporcion_largo["origen_madre"].replace({
    "porcentaje_madres_internas": "madres chilenas",
    "porcentajes_madres_extranjeras": "madres extranjeras"})
# Orden de regiones de norte a sur
orden_norte_sur = [
    "arica_parinacota",
    "tarapaca",
    "antofagasta",
    "atacama",
    "coquimbo",
    "valparaiso",
    "metropolitana",
    "ohiggins",
    "maule",
    "nuble",
    "biobio",
    "la_araucania",
    "los_rios",
    "los_lagos",
    "aysen",
    "magallanes"
]

# Selector de año
selector_ano = alt.binding_select(
    options=sorted(df_proporcion_largo["ano"].unique().tolist()),
    name="Año: "
)

param_ano = alt.param(
    name="Seleccionar_anio",
    bind=selector_ano,
    value=int(df_proporcion_largo["ano"].max())
)

# Crear gráfico
grafico_proporcion = alt.Chart(df_proporcion_largo).mark_bar().encode(
    y=alt.Y(
        "region:N",
        sort=orden_norte_sur,
        title="Región"
    ),
    x=alt.X(
        "porcentaje:Q",
        stack="normalize",
        title="Distribución de nacimientos según origen de la madre"
    ),
    color=alt.Color(
        "origen_madre:N",
        title="Origen de la madre"
    ),
    tooltip=[
        alt.Tooltip("ano:O", title="Año"),
        alt.Tooltip("region:N", title="Región"),
        alt.Tooltip("origen_madre:N", title="Origen de la madre"),
        alt.Tooltip("porcentaje:Q", title="Porcentaje", format=".2f")
    ]
).add_params(
    param_ano
).transform_filter(
    alt.datum.ano == param_ano
).properties(
    title="Proporción de nacimientos según origen de la madre por región",
    width=850,
    height=520
).interactive()

grafico_proporcion
```
