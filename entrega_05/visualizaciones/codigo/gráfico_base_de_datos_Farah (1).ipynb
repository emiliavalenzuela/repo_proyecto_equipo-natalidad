```python
import pandas as pd
import altair as alt
```

```python
df = pd.read_csv("jadue_database_limpia (2).csv", sep=";")
```

```python
df.head()
```

```python
df = df.copy()

df["Año"] = pd.to_numeric(df["Año"], errors="coerce")

df["tasa-especifica-fecundidad"] = (
    df["tasa-especifica-fecundidad"]
    .astype(str)
    .str.replace(",", ".", regex=False)
)

df["tasa-especifica-fecundidad"] = pd.to_numeric(
    df["tasa-especifica-fecundidad"],
    errors="coerce"
)

df = df.dropna(subset=["Año", "grupo-edad-madre", "tasa-especifica-fecundidad"])

df.head()
```

```python
df["tipo_grupo"] = df["grupo-edad-madre"].apply(
    lambda x: "15-19 años" if x == "15-19" else "Otros rangos etarios"
)

df.head()
```

```python
base = alt.Chart(df).encode(
    x=alt.X(
        "Año:O",
        title="Año"
    ),
    y=alt.Y(
        "tasa-especifica-fecundidad:Q",
        title="Tasa específica de fecundidad"
    ),
    detail="grupo-edad-madre:N",
    tooltip=[
        alt.Tooltip("Año:O", title="Año"),
        alt.Tooltip("grupo-edad-madre:N", title="Grupo de edad"),
        alt.Tooltip("tasa-especifica-fecundidad:Q", title="Tasa", format=".1f")
    ]
)

lineas_otros = base.transform_filter(
    alt.datum["grupo-edad-madre"] != "15-19"
).mark_line(
    color="#BBD7E6",
    strokeWidth=2,
    opacity=0.55
)

linea_adolescente = base.transform_filter(
    alt.datum["grupo-edad-madre"] == "15-19"
).mark_line(
    color="#0D3B4E",
    strokeWidth=3.2
)

puntos_adolescente = base.transform_filter(
    alt.datum["grupo-edad-madre"] == "15-19"
).mark_circle(
    color="#0D3B4E",
    size=55
)

grafico_fecundidad_comparado = (
    lineas_otros +
    linea_adolescente +
    puntos_adolescente
).properties(
    title="Evolución de la fecundidad adolescente y otros grupos etarios en Chile (1992-2024)",
    width=850,
    height=500
)

grafico_fecundidad_comparado
```
